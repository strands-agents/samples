# Swarms as Tools

This notebook demonstrates how to use the Agent(config=) constructor to load an agent configuration that includes a swarm as a tool. This pattern allows you to create coordinator agents that can delegate complex multi-agent tasks to specialized research teams.

## Overview

In this example, we'll:
1. Load an agent configuration that includes a research team swarm as a tool
2. Use the Agent(config=) constructor to create the coordinator
3. Demonstrate how the coordinator delegates research tasks to the swarm

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import yaml

# Load the simple graph configuration
with open('./configs/swarms-as-tools.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
```

```python
# Load the research coordinator agent using AgentConfigLoader
from strands.experimental.config_loader.agent import AgentConfigLoader

loader = AgentConfigLoader()
research_coordinator = loader.load_agent(config)

print("Research coordinator agent loaded successfully!") 

print(f"Tools: {research_coordinator.tool_names}")
```

```python
# Test the swarm tool with a comprehensive research request
print("Testing research coordinator with comprehensive research request...\n")

research_request = """
I need comprehensive research on 'Quantum Computing Applications in Healthcare'. 
Please provide detailed analysis including:
- Current applications and use cases
- Future potential and emerging opportunities
- Technical challenges and limitations
- Key players and recent developments
- Timeline for practical implementation

This research will be used for a strategic investment decision.
"""

response = research_coordinator(research_request)

print("=== RESEARCH COORDINATOR RESPONSE ===")
print(response)
print("=" * 50)
```

```python
# Test with different research parameters - basic depth
print("Testing research coordinator with basic depth analysis...\n")

basic_request = """
Research 'Sustainable Energy Storage Solutions' with basic depth analysis. 
Focus on the most promising technologies and their commercial viability. 
I need this for a preliminary market assessment.
"""

response = research_coordinator(basic_request)

print("=== RESEARCH COORDINATOR RESPONSE ===")
print(response)
print("=" * 50)
```

## Key Takeaways

This notebook demonstrates how the AgentConfigLoader pattern with swarms-as-tools enables:

1. **Declarative Multi-Agent Configuration**: Define complex swarm workflows in YAML without code
2. **Automatic Swarm Tool Loading**: The swarm is automatically converted to a callable tool
3. **Intelligent Task Delegation**: The coordinator delegates research tasks to the specialized team
4. **Collaborative Processing**: Multiple agents work together through defined handoff strategies

The research coordinator agent automatically:
- Analyzes research requests and delegates to the research team swarm
- Coordinates the sequential workflow: Researcher → Analyst → Synthesizer
- Handles different research depths (basic, comprehensive, expert)
- Produces structured, high-quality research outputs
- Adapts output format based on requirements (report, summary, analysis)

The research team swarm automatically:
- **Researcher**: Gathers comprehensive information and data
- **Analyst**: Analyzes findings and identifies key insights
- **Synthesizer**: Combines insights into coherent, actionable reports

This pattern is particularly powerful for:
- Complex research and analysis tasks requiring multiple perspectives
- Strategic planning and market analysis
- Competitive intelligence and trend analysis
- Investment research and due diligence
- Any scenario requiring collaborative expertise and structured workflows
