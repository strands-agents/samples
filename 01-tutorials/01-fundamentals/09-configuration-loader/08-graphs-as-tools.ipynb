# Graphs as Tools

This notebook demonstrates how to use the AgentConfigLoader to load an agent configuration that includes a graph as a tool. This pattern allows you to create coordinator agents that can delegate complex sequential workflows to structured document processing pipelines.

## Overview

In this example, we'll:
1. Load an agent configuration that includes a document processing graph as a tool
2. Use the AgentConfigLoader to create the coordinator
3. Demonstrate how the coordinator delegates document processing tasks to the graph

````python
from dotenv import load_dotenv
load_dotenv()
````

````python
import yaml

# Load the graphs-as-tools configuration
with open('./configs/graphs-as-tools.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
````

````python
# Load the document processing coordinator agent using AgentConfigLoader
from strands.experimental.config_loader.agent import AgentConfigLoader

loader = AgentConfigLoader()
document_coordinator = loader.load_agent(config)

print("Document processing coordinator agent loaded successfully!")

print(f"Tools: {document_coordinator.tool_names}")
````

````python
# Define sample document for testing
healthcare_document = """
# Artificial Intelligence in Modern Healthcare

Artificial Intelligence (AI) is revolutionizing healthcare through various applications including 
diagnostic imaging, drug discovery, personalized treatment plans, and predictive analytics. 
Machine learning algorithms can now detect diseases earlier and more accurately than traditional methods.

## Key Applications

### Diagnostic Imaging
AI-powered imaging systems can identify tumors, fractures, and other abnormalities with remarkable 
precision. Deep learning models trained on millions of medical images can often spot patterns 
that human radiologists might miss.

### Drug Discovery
Traditional drug discovery takes 10-15 years and costs billions of dollars. AI is accelerating 
this process by predicting molecular behavior, identifying promising compounds, and optimizing 
clinical trial designs.

### Personalized Medicine
By analyzing genetic data, medical history, and lifestyle factors, AI can help create personalized 
treatment plans that are more effective and have fewer side effects.

## Key Benefits
- Improved diagnostic accuracy (up to 95% in some imaging applications)
- Reduced treatment costs (estimated 20-30% savings in some areas)
- Personalized patient care based on individual characteristics
- Accelerated drug development (reducing timelines by 2-3 years)
- Enhanced clinical decision support for healthcare providers

## Challenges and Considerations
- Data privacy and security concerns
- Regulatory approval processes
- Integration with existing healthcare systems
- Need for healthcare professional training
- Ensuring AI transparency and explainability

The future of healthcare will likely see even deeper integration of AI technologies, 
leading to more precise, efficient, and accessible medical care for patients worldwide.
"""

print("Healthcare document prepared for processing.")
print(f"Document length: {len(healthcare_document)} characters")
````

````python
# Test the graph tool with comprehensive document analysis
print("Testing document coordinator with comprehensive analysis...\n")

analysis_request = f"""
Please process this healthcare document and generate a comprehensive analysis:

{healthcare_document}

I need a detailed analysis that includes key themes, insights, and recommendations 
for healthcare technology investment decisions.
"""

response = document_coordinator(analysis_request)

print("=== DOCUMENT COORDINATOR RESPONSE ===")
print(response)
print("=" * 50)
````

## Key Takeaways

This notebook demonstrates how the AgentConfigLoader pattern with graphs-as-tools enables:

1. **Declarative Workflow Configuration**: Define complex sequential processing workflows in YAML without code
2. **Automatic Graph Tool Loading**: The graph is automatically converted to a callable tool
3. **Intelligent Task Delegation**: The coordinator delegates document processing tasks to the structured workflow
4. **Sequential Processing**: Documents flow through a structured pipeline with conditional logic

The document processing coordinator agent automatically:
- Analyzes document processing requests and delegates to the document processor graph
- Coordinates the sequential workflow: Validator → Analyzer → Processor → Synthesizer
- Handles different document types (healthcare, technical, business)
- Produces structured, high-quality analysis outputs
- Adapts processing based on document type and requested output format

The document processing graph automatically:
- **Validator**: Checks document format, quality, and classifies document type
- **Analyzer**: Extracts key themes, topics, and structural information
- **Processor**: Applies specialized processing based on document type and requirements
- **Synthesizer**: Combines all findings into coherent, actionable outputs

This pattern is particularly powerful for:
- Content analysis and document processing pipelines
- Quality assurance workflows with validation steps
- Multi-stage analysis requiring different specialized processing
- Document classification and routing systems
- Any scenario requiring structured, sequential processing with conditional logic
