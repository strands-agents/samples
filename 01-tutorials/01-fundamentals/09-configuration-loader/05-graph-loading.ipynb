# Loading Graphs from Configuration

The GraphConfigLoader enables you to create multi-agent graphs from YAML configuration files, making it easy to define, version, and manage complex agent workflows declaratively. The GraphBuilder.from_config() method provides a convenient interface to the underlying GraphConfigLoader.

## What is a Graph?

A Graph is a structured multi-agent system where agents are connected through explicit dependencies and conditions:

* **Directed workflow** - Agents execute in a specific order based on dependencies
* **Conditional routing** - Dynamic paths based on intermediate results
* **Parallel execution** - Independent branches can run concurrently
* **Deterministic flow** - Predictable execution patterns for complex workflows

## Configuration-Driven Graphs

Instead of building graphs programmatically, you can define your entire workflow in a YAML configuration file. This approach provides:

- **Declarative definition** - Define nodes, edges, and conditions in YAML
- **Version control** - Track changes to your workflow configurations
- **Easy deployment** - Load different graph configurations for different environments
- **Maintainability** - Modify workflow behavior without changing code

## Example 1: Simple Graph - Basic Processing

Let's start with a simple graph configuration that demonstrates basic processing with one coordinator leading two specialists:

```python
import yaml

# Load the simple graph configuration
with open('./configs/graph-simple.strands.yml', 'r') as file:
    simple_config = yaml.safe_load(file)
    
print("Simple Graph Configuration Structure:")
print(f"Graph ID: {simple_config['graph']['graph_id']}")
print(f"Nodes: {len(simple_config['graph']['nodes'])}")
print(f"Edges: {len(simple_config['graph']['edges'])}")
print(f"Entry Points: {simple_config['graph']['entry_points']}")
```

```python
from strands.experimental.config_loader.graph.graph_config_loader import GraphConfigLoader

# Load simple graph using GraphConfigLoader.load_graph() method
config_loader = GraphConfigLoader()
simple_graph = config_loader.load_graph(config=simple_config)

print("✅ Simple Graph Created Successfully")
print(f"📊 Total Nodes: {len(simple_graph.nodes)}")
print(f"🚀 Entry Points: {[node.node_id for node in simple_graph.entry_points]}")
print("\n🎯 Node Roles:")
for node_id, node in simple_graph.nodes.items():
    print(f"  - {node_id}: {getattr(node.executor, 'name', 'Agent')}")
```

```python
# Execute the simple graph
task = "Analyze the impact of remote work on employee productivity. Provide a short analysis, no need for follow ups."

print("Executing Simple Graph:")
print(f"Task: {task}")

result = simple_graph(task)

print("\nExecution Summary:")
print(f"Status: {result.status}")
print(f"Total nodes: {result.total_nodes}")
print(f"Completed nodes: {result.completed_nodes}")
print(f"Execution time: {result.execution_time}ms")

print("\nExecution Order:")
for node in result.execution_order:
    print(f"- {node.node_id}")
```

```python
# Display results from specific nodes
print("Expert Analysis:")
display(Markdown(str(result.results["expert"].result)))
```

## Example 3: Conditional Graph - Branching with Conditions

Finally, let's load a graph that demonstrates conditional routing based on classification:

```python
# Load the conditional graph configuration
with open('./configs/graph-conditional.strands.yml', 'r') as file:
    conditional_config = yaml.safe_load(file)
    
print("Conditional Graph Configuration Structure:")
print(f"Graph ID: {conditional_config['graph']['graph_id']}")
print(f"Nodes: {len(conditional_config['graph']['nodes'])}")
print(f"Edges: {len(conditional_config['graph']['edges'])}")
print(f"Entry Points: {conditional_config['graph']['entry_points']}")
print("\nConditional Edges:")
for edge in conditional_config['graph']['edges']:
    if edge.get('condition'):
        print(f"- {edge['from_node']} -> {edge['to_node']} (condition: {edge['condition']['function']})")
```

```python
# Load conditional graph using GraphConfigLoader.load_graph() method
config_loader = GraphConfigLoader()
conditional_graph = config_loader.load_graph(config=conditional_config)

print("✅ Conditional Graph Created Successfully")
print(f"📊 Total Nodes: {len(conditional_graph.nodes)}")
print(f"🚀 Entry Points: {[node.node_id for node in conditional_graph.entry_points]}")
print("\n🎯 Node Roles:")
for node_id, node in conditional_graph.nodes.items():
    print(f"  - {node_id}: {getattr(node.executor, 'name', 'Agent')}")
```

```python
# Execute the conditional graph with a technical request
technical_task = "Provide report on technical aspects of working from home, outline things to consider and key risk factors"

print("Executing Conditional Graph (Technical Request):")
print(f"Task: {technical_task}")

result = conditional_graph(technical_task)

print("\nExecution Summary:")
print(f"Status: {result.status}")
print(f"Total nodes: {result.total_nodes}")
print(f"Completed nodes: {result.completed_nodes}")
print(f"Execution time: {result.execution_time}ms")

print("\nExecution Order:")
for node in result.execution_order:
    print(f"- {node.node_id}")

print("\nClassifier Result:")
display(Markdown(str(result.results["classifier"].result)))

# Display the final result
if "technical_report" in result.results:
    print("\nTechnical Expert Report:")
    display(Markdown(str(result.results["technical_report"].result)))
elif "business_report" in result.results:
    print("\nBusiness Expert Report:")
    display(Markdown(str(result.results["business_report"].result)))
```

```python
# Execute the conditional graph with a business request
business_task = "Provide report on business impact of working from home, outline things to consider and key risk factors"

print("Executing Conditional Graph (Business Request):")
print(f"Task: {business_task}")

result = conditional_graph(business_task)

print("\nExecution Summary:")
print(f"Status: {result.status}")
print(f"Total nodes: {result.total_nodes}")
print(f"Completed nodes: {result.completed_nodes}")
print(f"Execution time: {result.execution_time}ms")

print("\nExecution Order:")
for node in result.execution_order:
    print(f"- {node.node_id}")

print("\nClassifier Result:")
display(Markdown(str(result.results["classifier"].result)))

# Display the final result
if "technical_report" in result.results:
    print("\nTechnical Expert Report:")
    display(Markdown(str(result.results["technical_report"].result)))
elif "business_report" in result.results:
    print("\nBusiness Expert Report:")
    display(Markdown(str(result.results["business_report"].result)))
```

## Key Benefits of Configuration-Driven Graphs

1. **Declarative Definition**: Define complex workflows in readable YAML files
2. **Version Control**: Track changes to workflow configurations over time
3. **Environment Flexibility**: Use different configurations for development, testing, and production
4. **Easy Maintenance**: Modify workflow behavior without changing application code
5. **Reproducibility**: Ensure consistent graph behavior across deployments
6. **Visual Clarity**: YAML structure makes workflow dependencies clear

## When to Use Configuration-Driven Graphs

- **Structured workflows** with clear dependencies and execution order
- **Conditional processing** where different paths are taken based on results
- **Parallel processing** where independent tasks can run concurrently
- **Production deployments** where configuration management is important
- **Complex decision trees** requiring multiple specialized agents
- **Audit trails** where execution order and dependencies must be tracked

## Graph vs Swarm: When to Use Which?

**Use Graphs when:**
- You need predictable, structured workflows
- Dependencies between agents are clear
- You want parallel processing capabilities
- Conditional routing is required

**Use Swarms when:**
- You need autonomous agent collaboration
- Agents should decide when to hand off to others
- The workflow is more exploratory and adaptive
- You want emergent behavior from agent interactions
