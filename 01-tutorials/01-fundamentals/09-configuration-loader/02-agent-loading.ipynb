# Loading Agents from Configuration Files

This notebook demonstrates how to create and configure Strands Agents using **YAML configuration files** and the **AgentConfigLoader**. This declarative approach provides several advantages:

- **Configuration-Driven Development**: Define agent behavior, tools, and models in external config files
- **Environment Flexibility**: Easily switch between different configurations for development, testing, and production
- **Maintainability**: Separate agent logic from configuration, making updates easier
- **Reusability**: Share and version control agent configurations independently


## Creating an Agent from Configuration

Let's examine how to define and load a weather agent using YAML configuration and the AgentConfigLoader.

### 1. Weather Agent Configuration:

The configuration file defines:
- **Model**: Specifies which LLM to use (Claude 3.7 Sonnet via Amazon Bedrock)
- **System Prompt**: Sets the agent's behavior and capabilities
- **Tools**: Lists the tools available to the agent (weather_tool.weather)

This creates a specialized weather assistant that can:
- Answer weather-related queries using the weather tool
- Perform simple calculations as specified in the system prompt
- Maintain consistent behavior across different environments

```python
import yaml

with open('./configs/weather-agent.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
print(config)
```

```python
from pathlib import Path
from strands.experimental.config_loader.agent import AgentConfigLoader

# Create the config loader
loader = AgentConfigLoader()

# Load agent from dictionary config
weather_agent = loader.load_agent(config)

print(weather_agent)
```

```python
response = weather_agent("What is the weather today?")
print(response)
```
