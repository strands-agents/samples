# Agents as Tools: Multi-Agent Orchestration from Configuration

This notebook demonstrates how to create **sophisticated multi-agent systems** using configuration files, where specialized agents are exposed as tools to an orchestrator agent. This pattern enables:

- **Specialized Agent Tools**: Define domain-specific agents (research, product recommendations, travel planning) as reusable tools
- **Intelligent Routing**: An orchestrator agent automatically selects the appropriate specialist based on user queries
- **Scalable Architecture**: Add new specialized agents without modifying existing code
- **Configuration-Driven Multi-Agent Systems**: Define complex agent hierarchies in YAML


## Multi-Agent Orchestration Pattern

Let's examine the agents-as-tools configuration that creates a sophisticated multi-agent system.

### 1. Orchestrator Agent with Specialized Agent Tools:

The configuration defines:
- **Orchestrator Agent**: Routes queries to appropriate specialist agents
- **Research Assistant**: Handles factual research queries with citations
- **Product Recommendation Assistant**: Provides personalized product suggestions
- **Trip Planning Assistant**: Creates detailed travel itineraries
- **Research & Summary Assistant**: Performs research and creates concise summaries
- **Traditional Tools**: Includes file_write for output generation

This creates an intelligent system where:
- The orchestrator analyzes user queries and selects the most appropriate specialist
- Each specialist agent has focused expertise and tailored system prompts
- Agents can be combined with traditional function tools seamlessly
- The entire system is defined declaratively in configuration

```python
import yaml

with open('./configs/agents-as-tools.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
print(config)
```

```python
from pathlib import Path
from strands.experimental.config_loader.agent import AgentConfigLoader

# Create the config loader
loader = AgentConfigLoader()

# Load agent from dictionary config
orchestrator = loader.load_agent(config)

print(orchestrator)

print("Orchestrator agent loaded successfully!")

print(f"Tools: {orchestrator.tool_names}")
```

```python
# Example: E-commerce Customer Service System
import os

customer_query = (
    "I'm looking for hiking boots. Write the final response to current directory."
)

os.environ["BYPASS_TOOL_CONSENT"] = "true"

# The orchestrator automatically determines this requires multiple specialized agents
response = orchestrator(customer_query)
```
