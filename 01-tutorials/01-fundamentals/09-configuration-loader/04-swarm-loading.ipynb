# Loading Swarms from Configuration

The SwarmConfigLoader enables you to create multi-agent swarms from YAML configuration files, making it easy to define, version, and manage complex agent teams declaratively.

## What is a Swarm?

A Swarm is a collaborative multi-agent system where specialized agents work together autonomously to solve complex tasks through:

* **Autonomous coordination** - Agents decide when to hand off to other specialists
* **Shared context** - All agents have access to the full conversation history
* **Specialized roles** - Each agent has distinct expertise and capabilities
* **Emergent intelligence** - The team achieves better results than individual agents

## Configuration-Driven Swarms

Instead of creating agents programmatically, you can define your entire swarm team in a YAML configuration file. This approach provides:

- **Declarative definition** - Define agent roles, prompts, and parameters in YAML
- **Version control** - Track changes to your agent team configurations
- **Easy deployment** - Load different swarm configurations for different environments
- **Maintainability** - Modify agent behavior without changing code

## Examining the Swarm Configuration

Let's first look at our swarm configuration file to understand how agents are defined:

```python
import yaml

# Load the swarm configuration
with open('./configs/swarm.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
    
print("Swarm Configuration Structure:")
print(f"Agents: {len(config['swarm']['agents'])}")
print(f"Max Handoffs: {config['swarm']['max_handoffs']}")
print(f"Execution Timeout: {config['swarm']['execution_timeout']}s")
```

## Loading a Swarm from Configuration

The simplest way to create a swarm from configuration is using the SwarmConfigLoader:

```python
from strands.experimental.config_loader.swarm import SwarmConfigLoader

# Create the config loader
loader = SwarmConfigLoader()

# Load from dictionary config
swarm = loader.load_swarm(config)
```

```python
print("Swarm Created Successfully")
print(f"Total Agents: {len(swarm.nodes)}")
print(f"Max Handoffs: {swarm.max_handoffs}")
print(f"Max Iterations: {swarm.max_iterations}")
print(f"Node Timeout: {swarm.node_timeout}s")
print(f"Execution Timeout: {swarm.execution_timeout}s")
print(f"Repetetive Handoff Detection Window: {swarm.repetitive_handoff_detection_window}")
print(f"Repetetive Handoff Min Unique Agents: {swarm.repetitive_handoff_min_unique_agents}")
print("\nAgent Roles:")
for key, value in swarm.nodes.items():
    print(f"- {key}: {value.executor.description}")
```

## Executing the Swarm

Now let's put our swarm to work on a collaborative task that benefits from multiple perspectives:

```python
# Execute the swarm on a complex task
task = "Create a blog post explaining Agentic AI and then create a summary for a social media post."

print("Executing Task:")
print(f"Task: {task}")

result = swarm(task)
```

## Analyzing Swarm Results

Let's examine how the agents collaborated and what they accomplished:

```python
# Display execution summary
print("Execution Summary:")
print(f"Status: {result.status}")
print(f"Total Iterations: {result.execution_count}")
print(f"Execution Time: {result.execution_time}ms")
print(f"Tokens Used: {result.accumulated_usage['totalTokens']}")

print("\nAgent Collaboration Flow:")
for i, node in enumerate(result.node_history, 1):
    print(f"{i}. {node.node_id}")
```

```python
# Final Swarm Output
for agent_name in result.results.keys():
    print(f"- {agent_name}")
    display(Markdown(str(result.results[agent_name].result)))
```

## Key Benefits of Configuration-Driven Swarms

1. **Declarative Definition**: Define complex agent teams in readable YAML files
2. **Version Control**: Track changes to agent configurations over time
3. **Environment Flexibility**: Use different configurations for development, testing, and production
4. **Easy Maintenance**: Modify agent behavior without changing application code
5. **Reproducibility**: Ensure consistent swarm behavior across deployments

## When to Use Configuration-Driven Swarms

- **Complex multi-step tasks** requiring diverse expertise
- **Content creation workflows** needing research, creativity, and review
- **Analysis projects** benefiting from multiple perspectives
- **Production deployments** where configuration management is important
- **Team collaboration** where non-developers need to modify agent behavior
