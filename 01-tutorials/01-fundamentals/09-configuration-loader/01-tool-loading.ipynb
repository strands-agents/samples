# Loading Tools from Configuration Files

This notebook demonstrates how to use the **ToolConfigLoader** to dynamically load and instantiate tools from YAML configuration files. This approach enables:

- **Declarative Tool Management**: Define tools in configuration files rather than hardcoding them
- **Dynamic Tool Loading**: Load tools at runtime based on configuration
- **Flexible Tool Composition**: Mix custom tools with pre-built tools from strands-tools

## What You'll Learn

1. How to define tools in YAML configuration files
2. Using ToolConfigLoader to load custom and pre-built tools
3. Executing loaded tools with proper parameters
4. Best practices for tool configuration management

## Prerequisites

- Python 3.10 or later
- strands-agents and strands-agents-tools packages
- Basic understanding of YAML configuration files

```python
import yaml
import sys

with open('./configs/tools.strands.yml', 'r') as file:
    config = yaml.safe_load(file)
print(config)
```

```python
from strands.experimental.config_loader.tools.tool_config_loader import ToolConfigLoader

tool_loader = ToolConfigLoader()
weather = tool_loader.load_tool(tool=config["tools"][0])

print(weather)
```

## Demonstrating the file_write Tool

Now let's use the `file_write` tool to create a simple text file. This demonstrates how to:

1. Create a proper `ToolUse` request with the required parameters
2. Execute the tool asynchronously using the `stream` method
3. Handle the tool's response and verify the operation

The `file_write` tool requires two parameters:
- `path`: The file path where content should be written
- `content`: The text content to write to the file

```python
import os

# Set environment variable to bypass interactive prompts
os.environ['BYPASS_TOOL_CONSENT'] = 'true'

# Create a ToolUse request for the file_write tool
tool_use = {
    'toolUseId': 'demo-file-write',
    'name': 'file_write',
    'input': {
        'path': 'hello-strands.txt',
        'content': 'Hello Strands!'
    }
}

[result async for result in file_write.stream(tool_use, {})]
```
