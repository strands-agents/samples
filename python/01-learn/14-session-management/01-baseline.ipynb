# Session Management — Baseline (No Persistence)

This notebook demonstrates the default behavior of a Strands agent: conversation state lives only in memory.
When the kernel restarts (or the script ends), all messages are lost.

This establishes the problem that session managers solve.

```python
# Install dependencies
%pip install -q --upgrade strands-agents
```

## Create an agent without session management

```python
from strands import Agent

agent = Agent(
    system_prompt="You are a helpful assistant. Remember details the user shares with you."
)

# Tell the agent something memorable
response = agent("My favorite programming language is Python and I live in Seattle.")
print(response)
```

```python
# The agent remembers within the same session
response = agent("What's my favorite programming language?")
print(response)
```

## Simulate a restart

Creating a new agent instance simulates what happens when the kernel restarts or the process ends.
The conversation history is gone.

```python
# Create a fresh agent — simulating a restart
agent = Agent(
    system_prompt="You are a helpful assistant. Remember details the user shares with you."
)

# The agent has no memory of the previous conversation
response = agent("What's my favorite programming language?")
print(response)
```

The agent can't answer because the conversation history was lost.

In the next notebook, we'll fix this with `FileSessionManager`.
